```python
import os
import json
import random
import warnings
import joblib
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    PolynomialFeatures
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

warnings.filterwarnings("ignore")

# Reproducibility
SEED = 42
np.random.seed(SEED)
random.seed(SEED)

# Plot settings
sns.set_theme(
    style="whitegrid",
    context="notebook"
)

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 15
plt.rcParams["axes.labelsize"] = 12

print("Setup complete.")
```

We start by loading the classic Iris dataset and putting it into a tidy dataframe to work with.


```python
iris = load_iris(as_frame=True)

iris_df = iris.frame.copy()

iris_df["species"] = iris_df["target"].map(
    dict(enumerate(iris.target_names))
)

iris_df.drop(columns="target", inplace=True)

iris_df.head()
```

```python
print("Shape:", iris_df.shape)
print("\nColumns:")
print(iris_df.columns.tolist())

iris_df.info()
```

```python
iris_df.describe().T
```

Next, let's confirm how the three species are distributed across the dataset.


```python
plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=iris_df,
    x="species",
    hue="species",
    legend=False
)

plt.title("Iris Class Distribution")
plt.xlabel("Species")
plt.ylabel("Number of Samples")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()
```

### Observation

The Iris dataset contains three classes: *setosa*, *versicolor*,
and *virginica*. Each class contains the same number of observations,
so the dataset is balanced with respect to the target variable.

Therefore, accuracy is a reasonable primary evaluation metric for this
initial classification experiment.

With the classes confirmed as balanced, we can look at how each individual feature is distributed by species.


```python
features = iris.feature_names

fig, axes = plt.subplots(2, 2, figsize=(14, 9))

for ax, feature in zip(axes.flat, features):
    sns.histplot(
        data=iris_df,
        x=feature,
        hue="species",
        kde=True,
        element="step",
        ax=ax
    )

    ax.set_title(f"Distribution of {feature}")

plt.suptitle(
    "Feature Distributions Across Iris Species",
    fontsize=18,
    y=1.02
)

plt.tight_layout()
plt.show()
```

A pairplot lets us see every pairwise feature relationship at once, broken out by species.


```python
g = sns.pairplot(
    iris_df,
    hue="species",
    diag_kind="kde",
    corner=True,
    height=2.8
)

g.fig.suptitle(
    "Pairwise Relationships Between Iris Features",
    y=1.02,
    fontsize=18
)

plt.show()
```

### Interpretation

The pairplot reveals that **petal length** and **petal width** provide
strong separation between the three species.

*Setosa* forms a clearly separated cluster, while *versicolor* and
*virginica* show comparatively greater overlap.

This suggests that the petal-related features will be particularly
useful for classification.

To quantify these relationships more precisely, we compute a correlation heatmap of the numeric features.


```python
numeric_features = iris_df[features]

corr = numeric_features.corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

### Observation

Petal length and petal width exhibit a strong positive correlation.
Sepal length also shows a moderate positive relationship with the petal
measurements.

Although correlated features can introduce redundancy, Logistic Regression
can still perform effectively on this small dataset. Therefore, no feature
is removed at this stage; the decision is documented rather than made
solely from correlation magnitude.

## Preparing the data

Before modeling, we split the data into training and test sets.


```python
X = iris_df[features]
y = iris_df["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=SEED
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

### Why stratification?

The target contains three classes. Stratified sampling preserves
approximately the same class proportions in both training and testing
sets.

A fixed `random_state` ensures that the split can be reproduced.

## KNN

We start modeling with KNN, a simple distance-based classifier. Because KNN relies on distances between points, we scale the features first.


```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
k_values = [1, 3, 5, 7]

knn_results = []

for k in k_values:

    model = KNeighborsClassifier(n_neighbors=k)

    model.fit(X_train_scaled, y_train)

    predictions = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, predictions)

    knn_results.append({
        "k": k,
        "accuracy": accuracy
    })

knn_results = pd.DataFrame(knn_results)

knn_results
```

Let's visualize how test accuracy changes across the different values of k.


```python
plt.figure(figsize=(9, 5))

sns.lineplot(
    data=knn_results,
    x="k",
    y="accuracy",
    marker="o",
    linewidth=2.5
)

plt.xticks(k_values)
plt.ylim(0, 1.05)

plt.title("KNN Accuracy for Different Values of k")
plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Test Accuracy")

plt.tight_layout()
plt.show()
```

### KNN Interpretation

KNN classifies an observation based on the classes of nearby observations
in feature space. Standardization is applied because KNN relies on
distance calculations; otherwise, features with larger numerical scales
could disproportionately influence the distance.

The experiment evaluates the requested values of `k = 1, 3, 5, 7`.

## Logistic Regression

Next, we move to Logistic Regression, which will also serve as our final classifier for this dataset. We wrap scaling and the classifier together in a single pipeline, so preprocessing and modeling travel as one unit.



```python
logistic_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=SEED
        )
    )
])

logistic_pipeline.fit(X_train, y_train)
```

With the pipeline trained, we generate predictions on the held-out test set.


```python
y_pred = logistic_pipeline.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Test Accuracy: {accuracy:.4f}")
```

A classification report breaks performance down by class.


```python
print(classification_report(y_test, y_pred))
```

A confusion matrix gives a visual summary of correct vs. incorrect predictions.


```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=logistic_pipeline.classes_
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=logistic_pipeline.classes_,
    yticklabels=logistic_pipeline.classes_
)

plt.title("Logistic Regression — Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.tight_layout()
plt.show()
```

### Confusion Matrix Interpretation

The confusion matrix shows the number of correctly and incorrectly
classified samples for each Iris species.

Because the dataset is relatively small and well-separated, Logistic
Regression is expected to achieve high classification performance.
Any errors are primarily expected between *versicolor* and *virginica*,
whose feature distributions overlap more strongly than those of *setosa*. 
		Beyond hard class labels, Logistic Regression also produces class probabilities, which are worth a closer look.

```python
probabilities = logistic_pipeline.predict_proba(X_test)

probability_df = pd.DataFrame(
    probabilities,
    columns=logistic_pipeline.classes_
)

probability_df["Actual"] = y_test.values
probability_df["Predicted"] = y_pred

probability_df.head(10)
```

```python
sample_idx = np.arange(len(probabilities))

plt.figure(figsize=(14, 6))

for i, species in enumerate(logistic_pipeline.classes_):
    plt.plot(
        sample_idx,
        probabilities[:, i],
        marker="o",
        label=species
    )

plt.xlabel("Test Sample")
plt.ylabel("Predicted Probability")
plt.title("Logistic Regression Class Probabilities")
plt.legend()

plt.tight_layout()
plt.show()
```

### Probability Interpretation

Logistic Regression does not simply produce a class label. It estimates
the probability of an observation belonging to each class.

For a given sample, the predicted class corresponds to the class with
the highest estimated probability.

For example, if the model produces:

- Setosa: 0.01
- Versicolor: 0.91
- Virginica: 0.08

the predicted class is *versicolor* because it has the highest probability.

The probabilities also provide an indication of classification confidence.

We can also gauge overall model confidence by examining each sample's maximum predicted probability.


```python
confidence = probabilities.max(axis=1)

plt.figure(figsize=(10, 5))

sns.histplot(
    confidence,
    bins=10,
    kde=True
)

plt.title("Distribution of Model Prediction Confidence")
plt.xlabel("Maximum Predicted Probability")
plt.ylabel("Number of Test Samples")

plt.tight_layout()
plt.show()
```

### Observation

Most predictions have relatively high maximum class probabilities,
indicating that the model is confident in its predictions.

Lower-confidence predictions generally correspond to samples located
near the decision boundary between overlapping classes.

# Part Two: The Titanic Dataset

Having worked through the relatively clean Iris dataset, we now turn to a messier, real-world dataset: Titanic passenger records.


## Reproducible Data Ingestion

The Titanic dataset is loaded directly from a publicly accessible
GitHub repository rather than from a local file. This makes the notebook independent of the local Colab filesystem
and allows the data ingestion step to be reproduced from a fresh
runtime.

```python
TITANIC_URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

titanic = pd.read_csv(TITANIC_URL)

titanic.head()
```

```python
print("Shape:", titanic.shape)

titanic.info()
```

```python
titanic.describe(include="all").T
```

## Missing Values

Real-world data usually comes with gaps. Let's find out where they are.


```python
missing = titanic.isna().sum().sort_values(ascending=False)

missing[missing > 0]
```

```python
missing_table = pd.DataFrame({
    "Missing Count": titanic.isna().sum(),
    "Missing %": titanic.isna().mean() * 100
})

missing_table = missing_table[
    missing_table["Missing Count"] > 0
].sort_values("Missing %", ascending=False)

missing_table
```

A heatmap makes the pattern of missingness easy to see at a glance.


```python
plt.figure(figsize=(14, 7))

sns.heatmap(
    titanic.isna(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Titanic Dataset — Missing Value Map")
plt.xlabel("Features")
plt.ylabel("Records")

plt.tight_layout()
plt.show()
```

## Outliers

Next, we check the `Fare` column for extreme values.


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(
    titanic["Fare"],
    kde=True,
    ax=axes[0]
)

axes[0].set_title("Fare Distribution")
axes[0].set_xlabel("Fare")

sns.boxplot(
    x=titanic["Fare"],
    ax=axes[1]
)

axes[1].set_title("Fare Boxplot")
axes[1].set_xlabel("Fare")

plt.tight_layout()
plt.show()
```

```python
Q1 = titanic["Fare"].quantile(0.25)
Q3 = titanic["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = titanic[
    (titanic["Fare"] < lower_bound) |
    (titanic["Fare"] > upper_bound)
]

print("Number of Fare outliers:", len(fare_outliers))
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

### Outlier Decision — Fare

The `Fare` feature contains high-value observations that are identified
as statistical outliers using the IQR method.

These observations are not automatically removed because a high fare may
represent a legitimate passenger characteristic rather than an erroneous
measurement.

For this analysis, extreme values are handled using **IQR-based clipping**
rather than row deletion. This preserves potentially informative records
while reducing the influence of extreme values.

```python
titanic_clean = titanic.copy()

titanic_clean["Fare"] = titanic_clean["Fare"].clip(
    lower=lower_bound,
    upper=upper_bound
)
```

## Feature Engineering

With missing values and outliers addressed, we can engineer a few new features.


**FamilySize** combines siblings/spouses and parents/children aboard into a single family-size count.


```python
titanic_clean["FamilySize"] = (
    titanic_clean["SibSp"] +
    titanic_clean["Parch"] +
    1
)
```

**IsAlone** flags passengers who were travelling by themselves.


```python
titanic_clean["IsAlone"] = (
    titanic_clean["FamilySize"] == 1
).astype(int)
```

**Title** is extracted from each passenger's name, which often encodes useful social information (e.g. 'Mr', 'Mrs', 'Master').


```python
titanic_clean["Title"] = (
    titanic_clean["Name"]
    .str.extract(r",\s*([^.]*)\.")[0]
    .str.strip()
)
```

```python
#Group Rare Titles
common_titles = [
    "Mr", "Miss", "Mrs", "Master"
]

titanic_clean["Title"] = titanic_clean["Title"].where(
    titanic_clean["Title"].isin(common_titles),
    "Rare"
)
```

With the new features in place, we revisit the correlation heatmap.


```python
plt.figure(figsize=(12, 8))

corr = titanic_clean.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Titanic Numerical Feature Correlations")

plt.tight_layout()
plt.show()
```

### Feature Selection Decision

`PassengerId` is an identifier rather than a meaningful predictive
feature and therefore will not be used as a model input.

`SibSp` and `Parch` are retained initially because they represent
different aspects of family relationships, while `FamilySize` is added
as an engineered feature.

Feature removal is therefore based on both statistical redundancy
and semantic meaning rather than correlation alone.

## Building the Preprocessing Pipeline

We now assemble a full preprocessing pipeline for the Titanic features.


```python
#define features
target = "Survived"

features = [
    "Pclass",
    "Sex",
    "Age",
    "Fare",
    "Embarked",
    "FamilySize",
    "IsAlone",
    "Title"
]

X = titanic_clean[features]
y = titanic_clean[target]

#split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=SEED
)
```

```python
#Separate the numerical and categorical features
numeric_features = [
    "Pclass",
    "Age",
    "Fare",
    "FamilySize",
    "IsAlone"
]

categorical_features = [
    "Sex",
    "Embarked",
    "Title"
]

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median",
            add_indicator=True
        )
    ),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_pipeline,
        numeric_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])
```

We then attach a Logistic Regression classifier to the end of the pipeline.


```python
titanic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            random_state=SEED
        )
    )
])
```

Now we fit the full pipeline on the training data.


```python
titanic_pipeline.fit(
    X_train,
    y_train
)
```

...and evaluate it on the test set.


```python
titanic_predictions = titanic_pipeline.predict(X_test)

print(
    classification_report(
        y_test,
        titanic_predictions
    )
)
```

## Saving Artifacts

Finally, we persist the cleaned dataset, the fitted pipeline, and the train/test splits for reuse.


```python
output_dir = Path("/content/ml_lab_artifacts")
output_dir.mkdir(exist_ok=True)

titanic_clean.to_csv(
    output_dir / "clean_v1.csv",
    index=False
)

print("Saved:", output_dir / "clean_v1.csv")
```

The fitted pipeline is saved as well, so preprocessing can be reproduced later without refitting.


```python
pipeline_path = output_dir / "titanic_preprocessing_pipeline.joblib"

joblib.dump(
    titanic_pipeline,
    pipeline_path
)

print("Saved:", pipeline_path)
```

We save the train and test splits too, keeping the target column alongside the features.


```python
train_data = X_train.copy()
train_data[target] = y_train

test_data = X_test.copy()
test_data[target] = y_test

train_data.to_csv(
    output_dir / "train.csv",
    index=False
)

test_data.to_csv(
    output_dir / "test.csv",
    index=False
)
```

```python
X_test_transformed_original = (
    titanic_pipeline
    .named_steps["preprocessor"]
    .transform(X_test)
)

X_test_transformed_reloaded = (
    reloaded_pipeline
    .named_steps["preprocessor"]
    .transform(X_test)
)
```

```python
np.testing.assert_allclose(
    X_test_transformed_original,
    X_test_transformed_reloaded
)

print(" Preprocessing transformations are reproducible.")
```

```python
original_predictions = titanic_pipeline.predict(X_test)

reloaded_predictions = reloaded_pipeline.predict(X_test)

np.testing.assert_array_equal(
    original_predictions,
    reloaded_predictions
)

print("Reloaded pipeline produces identical predictions.")
print("Reproducibility verification successful.")
```
